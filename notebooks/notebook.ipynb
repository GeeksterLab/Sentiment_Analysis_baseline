#### Cleaning text

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
def clean_text(text: str) -> str:
    text = text.lower()
    text = re.sub(r'<[^>]+>', '', text) # delete HTML balise
    text = re.sub(r'http\S+', '', text) # delete URLs
    text = re.sub(r'[^a-z0-9\s]', ' ', text) # delete ponctuation
    text = re.sub(r'\s+', ' ', text) # delete duplicated spaces

    tokens = word_tokenize(text)
    stop_words = set(stopwords.words('english'))
    filtered_token = [word for word in tokens if word not in stop_words]
    text = ' '.join(filtered_token)

    return text

X = df["Text"]
y = df ["Feeling"]

cleaned = [clean_text(t) for t in X[:10]] # overview
for orig, clean in zip(X[:2], cleaned[:2]):
    print(f"BEFORE : {orig}")
    print(f"AFTER : {clean}\n")
```

```python
df["cleaned_text"] = [clean_text(t) for t in X]
df[["Text", "cleaned_text", "Feeling"]].head(5)
```

```python
from sklearn.utils.class_weight import compute_class_weight
from sklearn.model_selection import train_test_split
import numpy as np
```

```python
# ═════════════════════ Train Test Split ═════════════════════


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    stratify=y,
    test_size=0.2,
    random_state=42,
)
```

```python
# class_weight
class_weights = compute_class_weight('balanced', classes=np.array([0,1,2]), y=y_train)
class_weight_dict = {0: class_weights[0], 1: class_weights[1], 2: class_weights[2]}
```

### Tokenization

```python
from tf_keras.preprocessing.text import Tokenizer
from tf_keras.preprocessing.sequence import pad_sequences
import pickle
```

```python
# ═════════════════════ Tokenization + Embedding Keras ═════════════════════


# ░░░░░░░░░░░░░░ Params ░░░░░░░░░░░░░░
VOCAB_SIZE    = 20000           # Max vocab size
MAX_LEN       = 200            # Max sequence length
OOV_TOKEN     = "<OOV>"        # Token for unknown words
EMBEDDING_DIM = 100            # Embedding dimension

# ░░░░░░░░░░░░░░ Tokenizer ░░░░░░░░░░░░░░
tokenizer = Tokenizer(num_words=VOCAB_SIZE, oov_token=OOV_TOKEN)
tokenizer.fit_on_texts(X_train) # Learn vocab on the Train only
with open("../models/tokenizer_a.pkl", "wb") as f:
    pickle.dump(tokenizer, f)

word_index = tokenizer.word_index

# Intelligent check → vocab, OOV, frequency

print(f" Vocab length : {len(word_index)}")
print(f" Exemples : {list(word_index.items())[:10]}")
print(f" Frequency checking : {list(tokenizer.word_counts.items())[:10]}")

print(f" OOV Token index : {tokenizer.word_index.get(OOV_TOKEN)}")    # # Check if the token index has been created
print(f" Unknown word test : {tokenizer.texts_to_sequences(['this unknown_word is new'])}")    # Check his behavior

# ░░░░░░░░░░░░░░ Conversion texte → séquences ░░░░░░░░░░░░░░
X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq = tokenizer.texts_to_sequences(X_test)

print(f"\n Raw text : {X_train[0]}")
print(f"Tokenized sequency: {X_train_seq[0]}")

# ░░░░░░░░░░░░░░ Padding (même taille pour tous) ░░░░░░░░░░░░░░

X_train_pad = pad_sequences(
    X_train_seq,
    maxlen=MAX_LEN,
    padding='post',        # add 0 after the sequence
    truncating='post',     # cut is the sequence is too long
)

X_test_pad = pad_sequences(
    X_test_seq,
    maxlen=MAX_LEN,
    padding='post',
    truncating='post'
)

print(f"\nShape after padding (train): {X_train_pad.shape}")
print(f"Shape after padding (test) : {X_test_pad.shape}")
print(f" Padding sequences : {X_train_pad[0]}")
```

### WORD2VEC

```python
from gensim.models import Word2Vec
```

```python
# ═════════════════════ WORD2VEC GENSIM (mot → vecteur) → MATRICE D'EMBEDDING (Tokenization + Word2Vec Gensim) ═════════════════════

# ░░░░░░░░░░░░░░ Tokenizer word by word for Word2Vec ░░░░░░░░░░░░░░
# tokenized_sentences = [t.split() for t in X_train] # (raw text) w2v shoulb be train on `string` not on padded sequences

X_train_cleaned = df.loc[X_train.index, "cleaned_text"]
tokenized_sentences = [t.split() for t in X_train_cleaned]

# ░░░░░░░░░░░░░░ W2V ░░░░░░░░░░░░░░
w2v_model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=EMBEDDING_DIM,   # Vector size
    window=5,                    # context : 5 closest words
    min_count=1,                 # word = at least once
    workers=8,                   # parallelism
    sg=1,                        # 1=Skip-Gram | 0=CB0W
    epochs=10,
)

# Text: smiliar words
try:
    print(f"Similar word to 'movie':", w2v_model.wv.most_similar('movie', topn=5))
    print(f"'Movie' vector [:5] :", w2v_model.wv['movie'][:5])
    print(f"Words close to 'movie' vector : ", w2v_model.wv.similar_by_vector(w2v_model.wv['movie'], topn=5))
except KeyError:
    print("[INFO] Word not found in the Word2Vec dictionnary.")
except Exception:
    print("[INFO] Corpus to small for this exemple.")

# ░░░░░░░░░░░░░░ Créer la matrice d'embedding ░░░░░░░░░░░░░░
embedding_matrix = np.zeros((VOCAB_SIZE + 1, EMBEDDING_DIM))

for word, idx in tokenizer.word_index.items():
    if idx < VOCAB_SIZE and word in w2v_model.wv:
        embedding_matrix[idx] = w2v_model.wv[word]

print(f"\nEmbedding matrix shape : {embedding_matrix.shape}")
coverage = np.sum(np.any(embedding_matrix != 0, axis = 1))
print(f"Vocab coverage : {coverage}/{VOCAB_SIZE} words")
```

```python
import matplotlib.pyplot as plt

def plot_history(history, title):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

    ax1.plot(history.history['accuracy'], label='train')
    ax1.plot(history.history['val_accuracy'], label='val')
    ax1.set_title('Accuracy')
    ax1.legend()

    ax2.plot(history.history['loss'], label='train')
    ax2.plot(history.history['val_loss'], label='val')
    ax2.set_title('Loss')
    ax2.legend()

    fig.suptitle(title)
    plt.show()

MODEL_A = "../models/cnn_tokenization.keras"
MODEL_B = "../models/cnn_word2vec.keras"
MODEL = "../models/full_cnn_tokenization.keras"
```

#### Tokenization

```python
from tf_keras import layers, callbacks, regularizers, Sequential, optimizers, models
from tf_keras.callbacks import EarlyStopping, CSVLogger
from tf_keras.metrics import AUC, Recall, Precision
from tf_keras.layers import Embedding, Conv1D
from sklearn.utils import resample
```

```python
# ═════════════════════ TRAINING NLP simple 1 (Tokenization) ═════════════════════

def initialize_nlp_tokenization_simple_model(VOCAB_SIZE, EMBEDDING_DIM, MAX_LEN):
    baseline_1 = Sequential([
        layers.Input(shape=(MAX_LEN,)),      # ← padded text (NLP)
        layers.Embedding(
            input_dim=VOCAB_SIZE +1,
            output_dim=EMBEDDING_DIM,
            mask_zero=True,
            name = 'simple_embedding'
        ),

        layers.Conv1D(128, 5, activation='relu'),
        layers.GlobalMaxPool1D(),
        layers.Dense(64, activation='relu'),
        layers.Dense(3, activation='softmax')     # I have 3 classes

    ], name="baseline_nlp_simple_tokenization")

    return baseline_1

def compile_nlp_tokenization_simple_model(baseline_1):
    rmsprop = optimizers.legacy.RMSprop(learning_rate=1e-3)

    baseline_1.compile(
        optimizer=rmsprop,
        loss= 'sparse_categorical_crossentropy',
        metrics=["accuracy"]
    )

    return baseline_1

def build_nlp_tokenizaion_simple_model(VOCAB_SIZE, EMBEDDING_DIM, MAX_LEN):
    baseline_1 = initialize_nlp_tokenization_simple_model(VOCAB_SIZE, EMBEDDING_DIM, MAX_LEN)
    baseline_1 = compile_nlp_tokenization_simple_model(baseline_1)

    return baseline_1

def callbacks_nlp_tokenizer_simple():
    es_a = EarlyStopping(
        monitor='val_loss',
        patience=3,               # 2 or 4 for short epoch
        restore_best_weights=True,
        verbose=1
    )

    # Save the best version of our model after every epoch
    checkpoint_a = callbacks.ModelCheckpoint(
        MODEL_A,
        monitor='val_loss',
        verbose=1,
        save_best_only=True,
    )

    # Reduce the learning_rate if the model doesn't get better after a few epochs
    reduce_lr_a = callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=3,
        verbose=1,
        min_lr=1e-6,
    )

    return es_a, checkpoint_a, reduce_lr_a
```

#### WORD2VEC

```python
# ═════════════════════ TRAINING NLP simple 2 (Word2Vec) ═════════════════════

def initialize_nlp_word2vec_simple_model(VOCAB_SIZE, EMBEDDING_DIM, MAX_LEN, embedding_matrix):
    baseline_2 = Sequential([
        layers.Input(shape=(MAX_LEN)),
        layers.Embedding(
            input_dim= VOCAB_SIZE + 1,
            output_dim=EMBEDDING_DIM,
            weights=[embedding_matrix],
            mask_zero=True,
            trainable=True,
            name="baseline_nlp_simple_word2vec"
        ),

        layers.Conv1D(128,5, activation='relu'),
        layers.GlobalMaxPool1D(),
        layers.Dense(64, activation='relu'),
        layers.Dense(3, activation='softmax'),

    ], name="baseline_nlp_simple_word2vec")

    return baseline_2

def compile_nlp_word2vec_simple_model(baseline_2):
    rmsprop = optimizers.RMSprop(learning_rate=1e-3)

    baseline_2.compile(
        optimizer=rmsprop,
        loss='sparse_categorical_crossentropy',
        metrics=["accuracy"]
    )

    return baseline_2

def build_nlp_word2vec_simple_model(VOCAB_SIZE, EMBEDDING_DIM, MAX_LEN, embedding_matrix):
    baseline_2 = initialize_nlp_word2vec_simple_model(VOCAB_SIZE, EMBEDDING_DIM, MAX_LEN, embedding_matrix)
    baseline_2 = compile_nlp_word2vec_simple_model(baseline_2)

    return baseline_2

def callbacks_nlp_word2vec_simple():
    es_b = EarlyStopping(
        monitor='val_loss',
        patience=3,
        restore_best_weights=True,
        verbose=1,
    )

    checkpoint_b = callbacks.ModelCheckpoint(
        MODEL_B,
        monitor='val_loss',
        verbose=1,
        save_best_only=True,
    )

    reduce_lr_b = callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=3,
        verbose=1,
        min_lr=1e-6,
    )

    return es_b, checkpoint_b, reduce_lr_b
```

```python
baseline_2 = build_nlp_word2vec_simple_model(VOCAB_SIZE, EMBEDDING_DIM, MAX_LEN, embedding_matrix)
baseline_2.summary()

es_b, checkpoint_b, reduce_lr_b = callbacks_nlp_word2vec_simple()

history_cnn_w = baseline_2.fit(
    X_train_pad,
    y_train,
    epochs=50,
    batch_size=512,
    validation_split=0.1,
    callbacks=[es_b, checkpoint_b, reduce_lr_b],
    class_weight=class_weight_dict,
)

loss_cnn_w, acc_cnn_w = baseline_2.evaluate(X_test_pad, y_test, verbose=1)
print(f"\n📊 Test Accuracy (CNN) : {acc_cnn_w:.4f}")
print(f"📉 Test Loss (CNN) : {loss_cnn_w:.4f}")

plot_history(history_cnn_w, "CNN baseline B - Word2Vec")
```

```python
# Get the LLm
def get_llm():
    provider = os.getenv("LLM_PROVIDER")

    if provider == "openai":
        return OpenAI()

    else:
        raise ValueError(f"Unknown provider: {provider}")

def ask(prompt: str) -> str:
    provider = os.getenv("LLM_PROVIDER")
    model = os.getenv("LLM_MODEL")
    client = get_llm()

    if not model:
        raise ValueError("LLM_MODEL is not defined in your .env")

    if provider == "openai":
        message = client.chat.completions.create(
            model=model,
            max_tokens=100,
            messages=[
                {
                    "role": "system",
                    "content": f"""
                        "You are a sentiment classifier model.
                        "Return only one integer: 0, 1 or 2."
                    """,
                },
                {
                    "role": "user",
                    "content": prompt,
                },
            ],
        )

        return str((message.choices[0].message.content or "").strip())

    else:
        raise ValueError(f"Unknown provider: {provider}")

review = X_test.iloc[0]

print(ask(f"""
          Classify the sentiment of the following Amazon review.

            Return only one integer:
            0 = negative
            1 = neutral
            2 = positive

            Do not return any explanation.

            Review:
            {review}
        """,
    ))
```

### Final Conclusion

CNN tokenization achieves the same quality as an LLM, but:
- CNN → no cost, ~1 ms latency per review, runs locally
- LLM → per-token cost, ~1 s latency per review, relies on an external API

At 1 million reviews per day, the CNN adds €0 in marginal inference cost, while the LLM can cost thousands of euros.

When should you still choose an LLM?
- When we don’t have labeled data to train a model
- For highly specialized domain (such as medical or legal)
- When we want zero ML

```python
# ═════════════════════ TRAINING NLP FULL (Tokenization) ═════════════════════

def initialize_nlp_tokenization_model(VOCAB_SIZE, EMBEDDING_DIM, MAX_LEN):
    baseline = Sequential([
        layers.Input(shape=(MAX_LEN,)),      # ← padded text (NLP)
        layers.Embedding(
            input_dim=VOCAB_SIZE +1,
            output_dim=EMBEDDING_DIM,
            mask_zero=True,
            name = 'simple_embedding'
        ),

        layers.Conv1D(128, 5, activation='relu'),
        layers.GlobalMaxPool1D(),
        layers.Dense(64, activation='relu'),
        layers.Dense(3, activation='softmax')     # I have 3 classes

    ], name="baseline_nlp_tokenization")

    return baseline

def compile_nlp_tokenization_model(baseline):
    rmsprop = optimizers.legacy.RMSprop(learning_rate=1e-3)

    baseline.compile(
        optimizer=rmsprop,
        loss= 'sparse_categorical_crossentropy',
        metrics=["accuracy"]
    )

    return baseline

def build_nlp_tokenization_model_full(VOCAB_SIZE, EMBEDDING_DIM, MAX_LEN):
    baseline = initialize_nlp_tokenization_model(VOCAB_SIZE, EMBEDDING_DIM, MAX_LEN)
    baseline = compile_nlp_tokenization_model(baseline)

    return baseline

def callbacks_nlp_tokenizer_full():
    es_full = EarlyStopping(
        monitor='val_loss',
        patience=3,               # 2 or 4 for short epoch
        restore_best_weights=True,
        verbose=1
    )

    # Save the best version of our model after every epoch
    checkpoint_full = callbacks.ModelCheckpoint(
        MODEL,
        monitor='val_loss',
        verbose=1,
        save_best_only=True,
    )

    # Reduce the learning_rate if the model doesn't get better after a few epochs
    reduce_lr_full = callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=3,
        verbose=1,
        min_lr=1e-6,
    )

    return es_full, checkpoint_full, reduce_lr_full
```
